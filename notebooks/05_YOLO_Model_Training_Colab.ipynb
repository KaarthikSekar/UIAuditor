## Importing the YOLO Framework

Ultralytics provides the official implementation of the YOLO object detection framework.

The YOLO class is imported to load a pretrained model and begin transfer learning on the UI component dataset.

```python
!pip install ultralytics
```

```python
from ultralytics import YOLO
```

```python
import yaml

dataset_path = "/content/dataset"
data_yaml_path = f"{dataset_path}/data.yaml"

with open(data_yaml_path, "r") as f:
    data_config = yaml.safe_load(f)

data_config["path"] = dataset_path  # absolute path, fixes the Colab mismatch

with open(data_yaml_path, "w") as f:
    yaml.dump(data_config, f, sort_keys=False)

print(data_config)
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
    data=data_yaml_path,
    epochs=10,
    imgsz=640,
    batch=16,
    project="/content/drive/MyDrive/YOLO_Runs",
    name="rico_yolo_v1_quick",
)
```

```python
from ultralytics import YOLO

model = YOLO(
    "/content/drive/MyDrive/rico_yolo_v1_quick/weights/last.pt"
)
```

```python
model.train(
    data="/content/dataset/data.yaml",
    epochs=10,
    imgsz=640,
    batch=16,
    project="/content/drive/MyDrive/YOLO_Runs",
    name="rico_yolo_v2_continue"
)
```

```python
from ultralytics import YOLO

# Load the latest trained weights
model = YOLO("/content/drive/MyDrive/YOLO_Runs/rico_yolo_v2_continue/weights/last.pt")

# Fine-tune for 5 more epochs
model.train(
    data="/content/dataset/data.yaml",
    epochs=5,
    imgsz=640,
    batch=16,
    lr0=1e-4,                  # Lower learning rate for fine-tuning
    optimizer="AdamW",
    project="/content/drive/MyDrive/YOLO_Runs",
    name="rico_yolo_v3_continue"
)
```
