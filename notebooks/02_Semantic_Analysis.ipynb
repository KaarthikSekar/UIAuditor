```python
import json
from pathlib import Path

from PIL import Image
import matplotlib.pyplot as plt
```

```python
DATA_DIR = Path("../Data")

COMBINED_DIR = DATA_DIR / "combined"
SEMANTIC_DIR = DATA_DIR / "semantic_annotations"

print(COMBINED_DIR)
print(SEMANTIC_DIR)
print("Combined Exists:", COMBINED_DIR.exists())
print("Semantic Exists:", SEMANTIC_DIR.exists())
```

```python
semantic_images = list(SEMANTIC_DIR.glob("*.png"))
semantic_json = list(SEMANTIC_DIR.glob("*.json"))

print("Semantic Images :", len(semantic_images))
print("Semantic JSON   :", len(semantic_json))
```

```python
semantic_img = Image.open(semantic_images[0])

print("Filename :", semantic_images[0].name)
print("Size :", semantic_img.size)
print("Mode :", semantic_img.mode)

plt.figure(figsize=(4, 8))
plt.imshow(semantic_img)
plt.axis("off")
plt.show()
```
